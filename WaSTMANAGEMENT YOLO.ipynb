```python
import os
import shutil

# =========================================================
# YOUR EXISTING DATASET
# =========================================================

SOURCE = r"C:\Users\ASUS\Desktop\West Management\DATASET"

# =========================================================
# NEW YOLO DATASET
# =========================================================

YOLO_DATASET = r"C:\Users\ASUS\Desktop\West Management\YOLO_DATASET"

# =========================================================
# CLASSES
# =========================================================

CLASSES = {
    "bio": 0,
    "non_bio": 1
}

# =========================================================
# CREATE YOLO FOLDERS
# =========================================================

for split in ["train", "val", "test"]:

    os.makedirs(
        os.path.join(YOLO_DATASET, split, "images"),
        exist_ok=True
    )

    os.makedirs(
        os.path.join(YOLO_DATASET, split, "labels"),
        exist_ok=True
    )

# =========================================================
# PROCESS TRAIN / VAL / TEST
# =========================================================

for split in ["train", "val", "test"]:

    print("\n================================")
    print("Processing:", split)
    print("================================")

    for class_name, class_id in CLASSES.items():

        source_folder = os.path.join(
            SOURCE,
            split,
            class_name
        )

        if not os.path.exists(source_folder):

            print("❌ Folder not found:")
            print(source_folder)

            continue

        print("\nClass:", class_name)

        images = [
            f for f in os.listdir(source_folder)
            if f.lower().endswith(
                (".jpg", ".jpeg", ".png", ".bmp", ".webp")
            )
        ]

        print("Images:", len(images))

        for image_file in images:

            source_image = os.path.join(
                source_folder,
                image_file
            )

            # Make unique filename
            base_name = os.path.splitext(image_file)[0]

            new_name = (
                split + "_" +
                class_name + "_" +
                base_name
            )

            extension = os.path.splitext(
                image_file
            )[1]

            image_name = new_name + extension

            label_name = new_name + ".txt"

            destination_image = os.path.join(
                YOLO_DATASET,
                split,
                "images",
                image_name
            )

            destination_label = os.path.join(
                YOLO_DATASET,
                split,
                "labels",
                label_name
            )

            # -----------------------------------------
            # COPY IMAGE
            # -----------------------------------------

            shutil.copy2(
                source_image,
                destination_image
            )

            # -----------------------------------------
            # CREATE YOLO LABEL
            # -----------------------------------------
            #
            # One object occupies the whole image
            #
            # class x_center y_center width height
            #
            # -----------------------------------------

            with open(
                destination_label,
                "w"
            ) as f:

                f.write(
                    f"{class_id} 0.5 0.5 1.0 1.0\n"
                )

print("\n================================")
print("YOLO DATASET CREATED")
print("================================")

print("\nLocation:")
print(YOLO_DATASET)
```

```python
import os
import shutil

# =====================================================
# SOURCE DATASET
# =====================================================

SOURCE = r"C:\Users\ASUS\Desktop\West Management\DATASET"

# =====================================================
# YOLO DATASET
# =====================================================

YOLO_DATASET = r"C:\Users\ASUS\Desktop\West Management\YOLO_DATASET"

# =====================================================
# CLASSES
# =====================================================

CLASSES = {
    "bio": 0,
    "non_bio": 1
}

# =====================================================
# CREATE FOLDERS
# =====================================================

for split in ["train", "val", "test"]:

    os.makedirs(
        os.path.join(
            YOLO_DATASET,
            split,
            "images"
        ),
        exist_ok=True
    )

    os.makedirs(
        os.path.join(
            YOLO_DATASET,
            split,
            "labels"
        ),
        exist_ok=True
    )

# =====================================================
# COPY IMAGES + CREATE LABELS
# =====================================================

for split in ["train", "val", "test"]:

    print("\n==============================")
    print("Processing:", split)
    print("==============================")

    for class_name, class_id in CLASSES.items():

        source_folder = os.path.join(
            SOURCE,
            split,
            class_name
        )

        if not os.path.exists(source_folder):

            print("❌ Missing:", source_folder)

            continue

        files = [
            f for f in os.listdir(source_folder)
            if f.lower().endswith(
                (".jpg", ".jpeg", ".png", ".bmp", ".webp")
            )
        ]

        print(
            class_name,
            "images:",
            len(files)
        )

        for file in files:

            source_file = os.path.join(
                source_folder,
                file
            )

            name = os.path.splitext(file)[0]

            extension = os.path.splitext(file)[1]

            # Unique filename
            new_name = (
                f"{split}_{class_name}_{name}"
            )

            image_name = (
                new_name + extension
            )

            label_name = (
                new_name + ".txt"
            )

            destination_image = os.path.join(
                YOLO_DATASET,
                split,
                "images",
                image_name
            )

            destination_label = os.path.join(
                YOLO_DATASET,
                split,
                "labels",
                label_name
            )

            # Copy image
            shutil.copy2(
                source_file,
                destination_image
            )

            # Create YOLO label
            with open(
                destination_label,
                "w"
            ) as f:

                f.write(
                    f"{class_id} "
                    f"0.5 "
                    f"0.5 "
                    f"1.0 "
                    f"1.0\n"
                )

print("\n================================")
print("✅ CONVERSION COMPLETE")
print("================================")
```

```python
import cv2
import time
import os
from ultralytics import YOLO

# ==========================================
# FIND BEST MODEL
# ==========================================

SEARCH_PATH = r"C:\Users\ASUS\Desktop\West Management"

best_models = []

for root, dirs, files in os.walk(SEARCH_PATH):

    for file in files:

        if file.lower() == "best.pt":

            best_models.append(
                os.path.join(root, file)
            )

if not best_models:

    raise FileNotFoundError(
        "❌ best.pt नहीं मिला"
    )

# ==========================================
# SELECT MODEL
# ==========================================

BEST_MODEL = best_models[0]

print("================================")
print("MODEL")
print("================================")
print(BEST_MODEL)

model = YOLO(BEST_MODEL)

print("✅ Model loaded")

# ==========================================
# CAMERA
# ==========================================

cap = cv2.VideoCapture(0)

cap.set(
    cv2.CAP_PROP_FRAME_WIDTH,
    640
)

cap.set(
    cv2.CAP_PROP_FRAME_HEIGHT,
    480
)

if not cap.isOpened():

    print("❌ Camera could not be opened")

    raise SystemExit

print("================================")
print("LIVE CAMERA")
print("================================")
print("Press Q to quit")

previous_time = time.time()

# ==========================================
# CAMERA LOOP
# ==========================================

while True:

    ret, frame = cap.read()

    if not ret:

        print("❌ Camera frame error")

        break

    # ======================================
    # YOLO DETECTION
    # ======================================

    results = model.predict(

        source=frame,

        imgsz=320,

        conf=0.30,

        iou=0.50,

        device="cpu",

        max_det=50,

        verbose=False
    )

    result = results[0]

    # ======================================
    # DRAW ALL DETECTIONS
    # ======================================

    annotated_frame = result.plot()

    # ======================================
    # FPS
    # ======================================

    current_time = time.time()

    fps = 1 / max(
        current_time - previous_time,
        0.001
    )

    previous_time = current_time

    # ======================================
    # COUNT OBJECTS
    # ======================================

    total_objects = len(result.boxes)

    bio_count = 0
    non_bio_count = 0

    if result.boxes is not None:

        for box in result.boxes:

            class_id = int(box.cls[0])

            if class_id == 0:

                bio_count += 1

            elif class_id == 1:

                non_bio_count += 1

    # ======================================
    # DISPLAY FPS
    # ======================================

    cv2.putText(

        annotated_frame,

        f"FPS: {fps:.1f}",

        (20, 30),

        cv2.FONT_HERSHEY_SIMPLEX,

        0.7,

        (0, 255, 0),

        2
    )

    # ======================================
    # TOTAL OBJECTS
    # ======================================

    cv2.putText(

        annotated_frame,

        f"Total Objects: {total_objects}",

        (20, 60),

        cv2.FONT_HERSHEY_SIMPLEX,

        0.7,

        (0, 255, 0),

        2
    )

    # ======================================
    # BIO
    # ======================================

    cv2.putText(

        annotated_frame,

        f"Bio: {bio_count}",

        (20, 90),

        cv2.FONT_HERSHEY_SIMPLEX,

        0.7,

        (0, 255, 0),

        2
    )

    # ======================================
    # NON-BIO
    # ======================================

    cv2.putText(

        annotated_frame,

        f"Non-Bio: {non_bio_count}",

        (20, 120),

        cv2.FONT_HERSHEY_SIMPLEX,

        0.7,

        (0, 255, 0),

        2
    )

    # ======================================
    # SHOW CAMERA
    # ======================================

    cv2.imshow(

        "YOLO Bio Non-Bio Multi Object",

        annotated_frame
    )

    # ======================================
    # QUIT
    # ======================================

    if cv2.waitKey(1) & 0xFF == ord("q"):

        break

# ==========================================
# CLOSE CAMERA
# ==========================================

cap.release()

cv2.destroyAllWindows()

print("================================")
print("CAMERA CLOSED")
print("================================")
```
